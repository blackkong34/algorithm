#파이썬 출석수업과제

```
#5-(3) 월을 입력하면 계절을 표시하는 프로그램
#문자 Q를 입력하기 전까지 계속 프로그램이 실행될 수 있도록 while반복문작성
while(True) :
#사용자로부터 입력값을 받으며, 문자 Q를 받으면 종료될 수 있도록 정수로 변환하지 않는다.
    month_str = input("현재 월을 입력해 주세요 : (종료 Q) ")
#조건문으로 Q를 입력하면 프로그램이 종료되고 그 외 값은 정수값으로 변환한다.
    if (month_str == "Q") :
        break
    month = int(month_str)
#조건문으로 입력한 값에 따라서 계절을 출력할 수 있도록
#리스트를 사용하여 계절별 범위를 지정한다.
    if month in [3, 4, 5] :
        print("봄입니다.")
    elif month in [6, 7, 8] :
        print("여름입니다.")
    elif month in [9, 10, 11] :
        print("가을입니다.")
    elif month in [12, 1, 2] :
        print("겨울입니다.")
#잘못된 값이 입력되었을 경우 사용자에게 다시 입력할 것을 안내한다.
    else :
        print("잘못된 값을 입력했습니다. 다시 입력해주세요")
```
