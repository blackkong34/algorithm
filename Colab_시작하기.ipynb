#파이썬 출석수업과제

```
#5-(3) 월을 입력하면 계절을 표시하는 프로그램
#문자 Q를 입력하기 전까지 계속 프로그램이 실행될 수 있도록 while반복문작성
while(True) :
#사용자로부터 입력값을 받으며, 문자 Q를 받으면 종료될 수 있도록 정수로 변환하지 않는다.
    month_str = input("현재 월을 입력해 주세요 : (종료 Q) ")
#조건문으로 Q를 입력하면 프로그램이 종료되고 그 외 값은 정수값으로 변환한다.
    if (month_str == "Q") :
        break
    month = int(month_str)
#조건문으로 입력한 값에 따라서 계절을 출력할 수 있도록
#리스트를 사용하여 계절별 범위를 지정한다.
    if month in [3, 4, 5] :
        print("봄입니다.")
    elif month in [6, 7, 8] :
        print("여름입니다.")
    elif month in [9, 10, 11] :
        print("가을입니다.")
    elif month in [12, 1, 2] :
        print("겨울입니다.")
#잘못된 값이 입력되었을 경우 사용자에게 다시 입력할 것을 안내한다.
    else :
        print("잘못된 값을 입력했습니다. 다시 입력해주세요")
```

```
#5-(2) 구구단 문제
#올바른 값을 입력할 때까지 계속 값을 받기 위해서 while 반복문을 사용한다.
while (True) :
#사용자로부터 값을 입력받고 정수값으로 변환해준다.
    x = int(input("출력할 단을 입력하세요 : "))
#변수를 만들어 구구단의 결과값을 저장한다.
    gugudan = 0
#알맞은 값이 입력되었을 경우 구구단을 실행할 수 있도록 조건문을 만든다.
    if (1 <= x <= 9) :
#구구단이 입력한 단수까지만 계산할 수 있도록 반복문을 만든다.
        for i in range(1, x+1) :
#1부터 입력한 값까지 곱한 값을 gugudan 변수에 저장한다.
            gugudan = x * i
#1부터 1씩 증가하여 곱하는 구구단의 값을 출력한다.
            print(x,"x",i,"=",gugudan)
#i의 값이 조건값까지 도달했다면 반복문을 끝낼 수 있도록 break문을 작성한다.
        break
#알맞지 않은 값이 입력되었을 경우, 사용자에게 다시 입력할 수 있도록 안대한다.
    else :
        print("1에서 9까지의 값을 입력해주세요")
```

```
#5-(1)
#사용자로부터 값을 입력받는다. 양수만 입력할 수 있도록 안내한다.
x = int(input("양수를 입력해주세요 :"))
#조건문으로 입력받은 값이양수인지 확인한다.
if (x > 0) :
#결과값을 저장할 변수를 만든다.
    total = 0
#입력한 값까지 수식이 적용될 수 있도록 반복문(x+1)을 만든다.
    for i in range (1, x+1) :
#홀수인 경우 거듭제곱하여 더해준다.
        if (i % 2 == 1) :
            total += i ** 2
#짝수인 경우 거듭제곱하여 빼준다.
        else :
            total -= i ** 2
#잘못입력한 경우 오류임을 알린다.
else :
    print("양수만 입력이 가능합니다")


print(total)
```
